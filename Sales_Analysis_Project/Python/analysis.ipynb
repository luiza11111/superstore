```python
import pandas as pd
import io

file_name = next(iter(uploaded))
df = pd.read_csv(io.BytesIO(uploaded[file_name]), encoding="latin1")

print(df.head())
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_", regex=False)
)

print(df.columns.tolist())
```

```python
# Алғашқы 5 жол
display(df.head())

# Соңғы 5 жол
display(df.tail())

# Жолдар мен бағандар саны
print("Dataset Shape:", df.shape)

# Бағандар туралы ақпарат
df.info()

# Статистикалық ақпарат
display(df.describe())
```

```python
missing_values = df.isnull().sum()

print(missing_values)
```

```python
duplicates = df.duplicated().sum()

print("Duplicate Rows:", duplicates)
```

```python
# Дубликаттарды алып тастау
df = df.drop_duplicates().copy()

# Күндерді Date форматына өзгерту
df["order_date"] = pd.to_datetime(
    df["order_date"],
    errors="coerce"
)

df["ship_date"] = pd.to_datetime(
    df["ship_date"],
    errors="coerce"
)

# Сандық бағандарды тексеру
numeric_columns = [
    "sales",
    "profit",
    "quantity",
    "discount"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

# Негізгі талдау үшін қажет жолдарды тазалау
df = df.dropna(
    subset=["order_date", "sales", "profit"]
)

print("Cleaning completed!")
print("Final Shape:", df.shape)
```

```python
category_sales = (
    df.groupby("category")["sales"]
    .sum()
    .sort_values(ascending=False)
)

display(category_sales)

plt.figure(figsize=(8, 5))

category_sales.plot(kind="bar")

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)

plt.show()
```

```python
region_sales = (
    df.groupby("region")["sales"]
    .sum()
    .sort_values(ascending=False)
)

display(region_sales)

plt.figure(figsize=(8, 5))

region_sales.plot(kind="bar")

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")

plt.xticks(rotation=0)
plt.show()
```

```python
df["month"] = df["order_date"].dt.to_period("M")

monthly_sales = (
    df.groupby("month")["sales"]
    .sum()
)

plt.figure(figsize=(14, 6))

monthly_sales.plot(kind="line", marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

```python
top_products = (
    df.groupby("product_name")["sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

display(top_products)

plt.figure(figsize=(10, 6))

top_products.sort_values().plot(kind="barh")

plt.title("Top 10 Products by Sales")
plt.xlabel("Total Sales")

plt.show()
```

```python
top_profit = (
    df.groupby("product_name")["profit"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

display(top_profit)

plt.figure(figsize=(10, 6))

top_profit.sort_values().plot(kind="barh")

plt.title("Top 10 Products by Profit")
plt.xlabel("Total Profit")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="sales",
    y="profit",
    hue="category"
)

plt.title("Sales vs Profit")

plt.show()
```

```python
df.to_csv(
    "cleaned_data.csv",
    index=False
)

print("Cleaned dataset saved!")
```
